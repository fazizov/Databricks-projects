```python
from pyspark.sql.datasource import DataSource, DataSourceReader
```

```python
class bronze_reader(DataSourceReader):
  def __init__(self, table_name):
    self.table = table_name
  def read(self):
      return spark.read.table(self.table)

class bronze_writer(DataSourceReader):
  def __init__(self,options):
    self.options = options
    self.table_name=options['table']
  def write(self,df):
      df.write.saveAsTable(self.table_name)
      return
      
class bronze(DataSource):
  def __init__(self, table_name,options):
    self.table = table_name
    self.options = options
  def createReader(self):
    return bronze_reader(self.table)
  def createWriter(self):
    return bronze_writer(self.table,self.options)
  
```

```python
spark.dataSource.register(bronze)
df=spark.read.format("bronze").table('bronze.cdf_source')
display(df)
```
