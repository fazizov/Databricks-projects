```sql
USE cdc;
drop table if exists cdc_watermarks;
Create table  cdc_watermarks(
  table_name string,
  last_lsn  bigint
)
```

```sql
create or replace function get_last_lsn(tblname string) 
returns bigint 
RETURN (
select last_lsn from cdc.cdc_watermarks where table_name = tblname LIMIT 1)
```
